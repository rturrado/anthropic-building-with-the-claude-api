# System prompts

This exercise checks how a system prompt steers the assistant's style.
The system prompt `"You are a Python engineer who writes very concise code."` pushes Claude toward short,
idiomatic code with minimal commentary.


```python
from chat import add_user_message, chat

messages = []

add_user_message(messages, "Write a Python function that checks a string for duplicate characters.")
system = "You are a Python engineer who writes very concise code."
answer = chat(messages, system=system)
print(answer)
```
