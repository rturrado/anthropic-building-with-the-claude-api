# Controlling output: prefill + stop sequences

This exercise checks two techniques for pinning Claude's output format tightly:

- **Prefilled assistant message.**
  The assistant turn is seeded with a setup sentence that ends with a newline and the opening of a bash code fence.
  The sentence carries an extra constraint ("in a single block without any comments") that would feel out of place in the user prompt,
  and the trailing fence pins the format.
  Claude continues from exactly that point, so no preamble is possible.
- **Stop sequence.**
  `stop_sequences` is set to the closing code-fence marker, so generation ends the moment Claude emits it.
  The returned text is the body of the block, nothing more.

Together, these pin the output format without any parsing or post-processing.


````python
from chat import add_assistant_message, add_user_message, chat

messages = []

prompt = "Generate 3 different sample AWS CLI commands. Each should be very short"
add_user_message(messages, prompt)
add_assistant_message(messages, "Here are all 3 commands in a single block without any comments:\n```bash")
answer = chat(messages=messages, stop_sequences=["```"])
print(answer)
````
