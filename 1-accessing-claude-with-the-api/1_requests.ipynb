# Conversation context

This exercise checks that the conversation keeps context across turns.
Both user and assistant messages are stored in a running list,
so a follow-up like "And Socrates?" is answered against the previous exchange about Plato and Aristotle.


```python
from chat import add_assistant_message, add_user_message, chat

messages = []

while True:
    user_input = input("> ")
    print(">", user_input)
    add_user_message(messages, user_input)
    assistant_answer = chat(messages)
    add_assistant_message(messages, assistant_answer)
    print(assistant_answer)
```
