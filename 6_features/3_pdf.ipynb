# PDF support

Same mechanism as `2_images` but with a different content block type:
`document` instead of `image`, `application/pdf` instead of an image MIME.
`files_api.attach_file(path)` returns the right block shape automatically from the file extension.

## What Claude reads

Text, images, charts, tables, mixed content inside the PDF — all extracted in a single call,
no local pre-processing.
Up to 32MB per file.

Model: `claude-sonnet-4-6`.


```python
from chat_tools import add_user_message, chat, text_from_message
from files_api import attach_file
from models import MODEL_SONNET
```

```python
messages = []
pdf_block = attach_file("earth.pdf")
text_block = {"type": "text", "text": "Summarize the document in one sentence."}
add_user_message(messages, [pdf_block, text_block])

response = chat(messages, model=MODEL_SONNET, max_tokens=4000)
print(text_from_message(response))
```
