# Citations

Attach a document with `citations: {"enabled": true}`
and Claude will tie each cited sentence in its answer to the location in the source where it came from.
Great for traceability: the user can hover a sentence in the UI
and see page number + exact cited text.

## Citation types

- `citation_page_location`: for PDFs.
  Carries `document_index`, `document_title`, `start_page_number`, `end_page_number`, `cited_text`.
- `citation_char_location`: for plain-text documents. Carries character offsets inside the text block.

## Response shape

`response.content` is still a list of `text` blocks,
but each one may carry a `citations` array linking that sentence to its source.
The demo loops over blocks and prints the citation metadata alongside each cited sentence.

Model: `claude-sonnet-4-6`.


```python
from chat_tools import add_user_message, chat
from files_api import attach_file
from models import MODEL_SONNET
```

```python
messages = []
pdf_block = attach_file("earth.pdf", citations=True, title="earth.pdf")
text_block = {"type": "text", "text": "How were Earth's atmosphere and oceans formed?"}
add_user_message(messages, [pdf_block, text_block])

response = chat(messages, model=MODEL_SONNET, max_tokens=4000)

for block in response.content:
    print(f"--- {block.type} ---")
    if block.type == "text":
        print(block.text)
        for c in getattr(block, "citations", None) or []:
            print(f"  [cite] p.{c.start_page_number}-{c.end_page_number}: {c.cited_text[:120]}")
    else:
        print(str(block)[:300])
    print()
```
