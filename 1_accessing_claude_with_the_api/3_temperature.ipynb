# Temperature

This exercise checks how temperature affects output variability.
The same prompt is sampled 20 times at `temperature=0.0` and at `temperature=1.0`, counting unique answers.
At 0 the model is deterministic (always picks the top token at each step);
at 1 it samples from the full distribution.


```python
from collections import Counter

from chat import add_user_message, chat

messages = []
add_user_message(
    messages, "Pick a small, non-touristy town in Asturias. Reply with only the name of the town, nothing else."
)
system = "You are an expert in Spanish geography."

results = {}
for temperature in 0.0, 1.0:
    counts = Counter()
    for _ in range(20):
        answer = chat(messages, system=system, temperature=temperature).strip()
        counts[answer] += 1
    results[temperature] = counts

for temperature, counts in results.items():
    print(f"--- temperature={temperature} ---")
    print(f"  unique towns: {len(counts)} / 20")
    for town, n in counts.most_common():
        print(f"  {n:>3}  {town}")
    print()
```
